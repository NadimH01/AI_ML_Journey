```python
#univariate  analysis
sns.countplot(data=df,x=df['Survived'])
```

```python
(df['Survived'].value_counts() / len(df)) * 100
```

```python
survived_count=(df['Survived'].value_counts())
survived_labels=df['Survived'].unique()
print(survived_labels)
plt.pie(survived_count,labels=survived_labels,autopct="%1.1f")
plt.title("Pie chart of survival rate")
plt.show()
```

```python
#Pclass
sns.countplot(data=df,x=df['Pclass'])
```

# Numerical analysis

```python
sns.histplot(df['Age'])
plt.title('Histrogram of age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title('Kde plot')
plt.xlabel('Age')
plt.ylabel('Probability distribution')
plt.show()
```

```python
sns.histplot(df['Fare'])
plt.title('Histrogram of Fare')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.kdeplot(df['Fare'])
plt.title('Kde plot')
plt.xlabel('Fare')
plt.ylabel('Probability distribution')
plt.show()
```
