# 2d line plot from pandas

```python
df=pd.read_csv('enrollment_data.csv')
```

```python
df
```

```python
plt.xlabel('Years') 
plt.ylabel('Programming & Digital Marketing')
plt.title('Students enrollment over the year')
plt.plot(df['Year'],df['Programming']) 
plt.plot(df['Year'],df['Digital Marketing']) 
```

```python
#adding label using plt.legend() 
plt.xlabel('Years') 
plt.ylabel('Programming & Digital Marketing')
plt.title('Students enrollment over the year')
plt.plot(df['Year'],df['Programming'],label='Programming') 
plt.plot(df['Year'],df['Digital Marketing'],label ='Marketing') 
plt.legend()
```

```python
# Adding visualization  
plt.xlabel('Years') 
plt.ylabel('Programming & Digital Marketing')
plt.title('Students enrollment over the year')
plt.plot(df['Year'],df['Programming'],label='Programming',color='red') 
plt.plot(df['Year'],df['Digital Marketing'],label ='Marketing',color='#32a8a4') 
plt.legend()
```
