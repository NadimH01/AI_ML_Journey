Phase 2: Exploratory Data Analysis (EDA).

```python
# Missing values in each column
missing_values = df.isnull().sum()
print(missing_values)
```

```python
# Missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage.round(2)
})

print(missing_df.sort_values(by="Percentage (%)", ascending=False))
```

# Frequency Distribution (%)

```python
gender_percent=(df["Sex"].value_counts(normalize=True).mul(100).round(2))
gender_percent
```

# Survival Percentage

```python
survival_percent = (
    df["Survived"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
display(survival_percent)
```

# Identify Numerical and Categorical Features

```python
numerical_columns = df.select_dtypes(include=["number"]).columns
print(numerical_columns.tolist())
```

```python
categorical_columns = df.select_dtypes(include=["object"]).columns

print(categorical_columns.tolist())
```

# Summary

```python
eda_summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Unique Values": df.nunique(),
    "Missing (%)": ((df.isnull().sum() / len(df)) * 100).round(2)
})

display(eda_summary.sort_values("Missing (%)", ascending=False))
```
